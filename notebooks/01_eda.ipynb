```python
import json
import matplotlib.pyplot as plt

with open("../data/processed/eda_stats.json") as f:
    stats = json.load(f)

cat_freq = stats["category_frequency"]
counts = list(cat_freq.values())

print("Total distinct categories:", len(cat_freq))
print("Most common category:", list(cat_freq.items())[0])
print("Least common categories (bottom 10):", list(cat_freq.items())[-10:])
print("\nCategories with fewer than 100 papers:", sum(1 for c in counts if c < 100))
print("Categories with fewer than 500 papers:", sum(1 for c in counts if c < 500))
print("Categories with fewer than 1000 papers:", sum(1 for c in counts if c < 1000))
print("\nMedian category frequency:", sorted(counts)[len(counts)//2])
```

```python
labels_dist = stats["labels_per_paper_distribution"]
print("Labels per paper distribution:")
for k, v in labels_dist.items():
    print(f"  {k} label(s): {v} papers")

print("\nMean title word count:", round(stats["title_word_count_mean"], 2))
print("Mean abstract word count:", round(stats["abstract_word_count_mean"], 2))

print("\nTitle length buckets:", stats["title_length_buckets"])
print("\nAbstract length buckets:", stats["abstract_length_buckets"])
```
